```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
D_electronico = 20e-3
Potencia = 100
T_aluminio = 27
T_base = T_aluminio
T_aire = 27
i_alfa_contacto = 5e-5
Q = Potencia
Area = np.pi * (D_electronico/2)**2

R_contacto = i_alfa_contacto / Area

```

```python
#Calcule la temperatura que alcanzará el dispositivo suponiendo que toda la potencia que el mismo genera debe transferirse por conducción al bloque.


# Rcontacto = (T_electronico - T_base) * Area / Q
T_electronico = T_base + Q * R_contacto
print(f"La temperatura del dispositivo es {T_electronico:.2f} °C")
```

```python
h_c = 1000  # W/m-2K
R_conv = 1 / (h_c * Area)
Rtotal = (1/R_contacto + 1/R_conv)**-1
T_electronico_conv = T_base + Q * Rtotal
```
